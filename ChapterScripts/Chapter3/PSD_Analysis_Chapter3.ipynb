```python
def PSDanalysis_voltage(list2, list3, t, nperseg=int(1e6)):
    """
    Compute single-sided detector voltage PSD (V^2/Hz) from a time trace.

    Parameters
    ----------
    list2 : array-like
        Time-domain voltage samples or a 2D array whose first column is voltage [V].
    list3 : array-like
        Unused (kept for signature parity with the phase version).
    t : float
        Total duration of the trace [s]; fs = len(voltage) / t.
    nperseg : int
        Segment length for Welch PSD. Defaults to 1e6 (capped to len(voltage)).

    Returns
    -------
    f : ndarray
        Frequency axis [Hz].
    Svv : ndarray
        Voltage PSD in V^2/Hz (single-sided).
    """
    # Convert to numpy arrays and take first column if 2D, mirroring your pattern
    list2 = np.asarray(list2)
    list3 = np.asarray(list3)  # intentionally unused, kept for API symmetry

    voltage_vals = list2[:, 0] if list2.ndim > 1 else list2

    # Drop non-finite samples (NaN/Inf), count them like your print(count)
    finite_mask = np.isfinite(voltage_vals)
    count = np.sum(~finite_mask)
    voltage_vals = voltage_vals[finite_mask]

    # Sampling rate from length and total duration
    fs = len(voltage_vals) / float(t)

    # Detrend DC to avoid a giant 0-Hz spike (common practice for noise PSDs)
    voltage_vals = voltage_vals - np.mean(voltage_vals)

    # Ensure nperseg is sensible
    nperseg = int(min(nperseg, len(voltage_vals)))
    if nperseg < 8:
        raise ValueError("Not enough samples for Welch PSD (nperseg < 8).")

    # Welch PSD in V^2/Hz (single-sided)
    f, Svv = signal.welch(
        voltage_vals,
        fs=fs,
        window='hann',
        nperseg=nperseg,
        scaling='density',
        return_onesided=True
    )

    # Match your function's behavior: print a 'count'
    print(count)
    return f, Svv
```

```python
import numpy as np

def align_in_band(f_ref, S_ref, f_other, S_other, f1, f2):
    """
    Align S_other to S_ref by matching average level (in dB) over [f1, f2].
    Returns (S_other_scaled, dB_shift). Does NOT change the frequency axes.
    """
    # Make everything 1-D numpy arrays
    f_ref   = np.asarray(f_ref,   float).ravel()
    S_ref   = np.asarray(S_ref,   float).ravel()
    f_other = np.asarray(f_other, float).ravel()
    S_other = np.asarray(S_other, float).ravel()

    # Sort by frequency (just in case)
    iref   = np.argsort(f_ref);    f_ref,   S_ref   = f_ref[iref],   S_ref[iref]
    iother = np.argsort(f_other);  f_other, S_other = f_other[iother], S_other[iother]

    # Mask the alignment band on the reference grid
    m_ref = (f_ref >= f1) & (f_ref <= f2) & np.isfinite(S_ref) & (S_ref > 0)
    if not np.any(m_ref):
        raise ValueError("No reference points inside the chosen band.")

    # Interpolate the 'other' PSD onto the reference freq grid (in-band only)
    S_other_ib = np.interp(f_ref[m_ref], f_other, S_other, left=np.nan, right=np.nan)

    # Build a valid mask where both PSDs are finite/positive
    valid = np.isfinite(S_other_ib) & (S_other_ib > 0)
    if not np.any(valid):
        raise ValueError("No overlap after interpolation; check [f1,f2] and grids.")

    # Compute average dB offset over the overlap
    dB_ref   = 10*np.log10(S_ref[m_ref][valid])
    dB_other = 10*np.log10(S_other_ib[valid])
    dB_shift = np.nanmean(dB_ref - dB_other)

    # Scale the entire 'other' PSD so it matches the reference in-band
    scale = 10**(dB_shift/10.0)
    S_other_scaled = S_other * scale

    return S_other_scaled, dB_shift
```
