```python
#C5 DAY NO FIBRE - EM/NM
DNF_100_em_t, DNF_100_em_c = np.load('DNF_100_em.npy')
DNF_100_nm_t, DNF_100_nm_c = np.load('DNF_100_nm.npy')
D_PD_DB=np.load('day_long_pdiff_db.npy')
D_PD_ts=[]
for i in range(len(D_PD_DB)):
    D_PD_ts.append(i/1000)
plt.rcParams.update({'font.size': 10})
plt.rc('axes', titlesize=18)
plt.rc('axes', labelsize=16)

fig = plt.figure(figsize=(8, 6))

gs = fig.add_gridspec(
    nrows=2, ncols=2,
    height_ratios=[3.5, 1],
    width_ratios=[7, 1],
    hspace=0.02,   # smaller -> bottom axis closer
    wspace=0.01    # smaller -> histogram/legend column closer
)

ax_main = fig.add_subplot(gs[0, 0])
ax_sub  = fig.add_subplot(gs[1, 0], sharex=ax_main)
ax_hist = fig.add_subplot(gs[0, 1], sharey=ax_main)
ax_leg  = fig.add_subplot(gs[1, 1])

# ---------- Plot your data ----------
ax_main.plot(DNF_100_em_t, DNF_100_em_c*100, label='Emulation')
ax_main.plot(DNF_100_nm_t, DNF_100_nm_c*100, label='No Emulation')
ax_main.set_ylim(0, 50)
ax_main.set_xlim(0, 200)
#ax_main.spines['right'].set_visible(False)

ax_sub.plot(D_PD_ts, D_PD_DB)

# ---------- Labels / title ----------
ax_sub.set_xlabel('Time [$s$]')
ax_main.set_ylabel('QBER [$\%$]')
ax_sub.set_ylabel('$\Delta\eta$ [db]')
ax_main.set_title('QBER over time - 100Hz, Day (no fibre)')

# Hide top x tick labels (shared x)
plt.setp(ax_main.get_xticklabels(), visible=False)
ax_main.tick_params(axis='x', which='both', length=0)

# ---------- Histogram axis formatting ----------
ax_hist.set_xlabel('PD')
ax_hist.tick_params(axis='y', labelleft=False)
ax_hist.grid(False)
#ax_hist.spines['left'].set_visible(False)

ax_hist.tick_params(axis='y', which='both',
                    length=2, width=0.6, color='0.5')

ax_hist.minorticks_off()
fig.canvas.draw()  # ensure axis limits are finalized

ymin, ymax = ax_main.get_ylim()
bins = np.linspace(ymin, ymax, 200)

def add_hist(y, **kwargs):
    y = np.asarray(y)
    y = y[np.isfinite(y)]
    if y.size:
        ax_hist.hist(
            y,
            bins=bins,
            orientation='horizontal',
            histtype='step',
            **kwargs
        )

# Match colors to plotted lines
add_hist(DNF_100_em_c*100, color='C0', linewidth=1.5,density=True,)
add_hist(DNF_100_nm_c*100, color='C1', linewidth=1.5,density=True,)
# ---------- Legend panel ----------
ax_leg.axis("off")
handles, labels = ax_main.get_legend_handles_labels()
ax_hist.legend(handles, labels, loc="upper right", fontsize=14, frameon=True)

# ---------- Margins (do NOT touch hspace/wspace here) ----------
fig.subplots_adjust(left=0.12, right=0.95, bottom=0.10, top=0.92)

plt.show()
plt.savefig('C5_DayNF_100Hz_EMvsNM.pdf', dpi=300)
```
