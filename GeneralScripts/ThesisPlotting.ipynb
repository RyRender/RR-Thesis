```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import cumulative_trapezoid

def prepare_psd(f, S):
    f, S = np.asarray(f), np.asarray(S)
    valid = np.isfinite(f) & np.isfinite(S) & (f > 0) & (S > 0)
    f, S = f[valid], S[valid]
    order = np.argsort(f)
    f, S = f[order], S[order]
    if len(f) < 2 or np.any(np.diff(f) <= 0):
        raise ValueError("PSD frequencies must be distinct and increasing.")
    return f, S

on_f, on_Pxx = prepare_psd(on_f, on_Pxx)
off_f, off_Pxx = prepare_psd(off_f, off_Pxx)
det_f, det_Pxx = prepare_psd(det_f, det_Pxx)

f_low = max(1.0, on_f.min(), off_f.min(), det_f.min())
f_high = min(on_f.max(), off_f.max(), det_f.max())

# IMPORTANT: reduce f_high if the calibrated measurement bandwidth
# is smaller than the available array range.
f_lf = 12.0
if not f_low < f_lf < f_high:
    raise ValueError("The low-frequency band is outside the common PSD range.")

# Union of measured frequencies preserves narrow resolved features.
f = np.unique(np.concatenate([
    on_f, off_f, det_f, [f_low, f_lf, f_high]
]))
f = f[(f >= f_low) & (f <= f_high)]

def interpolate_psd(f_src, S_src):
    return np.exp(np.interp(np.log(f), np.log(f_src), np.log(S_src)))

S_on = interpolate_psd(on_f, on_Pxx)
S_off = interpolate_psd(off_f, off_Pxx)
S_det = interpolate_psd(det_f, det_Pxx)

# Hypothetical suppression of LF excess to the measurement floor.
# Do not increase bins already below the separately measured floor.
S_hyp = S_on.copy()
low_band = f <= f_lf
S_hyp[low_band] = np.minimum(S_on[low_band], S_det[low_band])

# Stay within the measured frequency range; avoid zero-width integration.
T = np.geomspace(2.0 / f_high, 1.0 / f_low, 500)

def variance_vs_time(S):
    # Integrate backwards to avoid subtracting nearly equal totals.
    tail = -cumulative_trapezoid(S[::-1], f[::-1], initial=0)[::-1]
    return np.interp(1.0 / T, f, tail)

def error_percent(var):
    return -50.0 * np.expm1(-var / 2.0)

series = [
    (S_off, "Stabilisation OFF", "tab:orange", "-"),
    (S_on, "Stabilisation ON", "tab:blue", "-"),
    (S_hyp, "Hypothetical LF suppression", "tab:purple", "--"),
    (S_det, "Detector-noise equivalent", "tab:green", ":"),
]

fig, axs = plt.subplots(2, 1, figsize=(8, 8),
                        constrained_layout=True)

for S, label, colour, style in series:
    axs[0].loglog(f, S, color=colour, ls=style, label=label)
    axs[1].semilogx(T, error_percent(variance_vs_time(S)),
                    color=colour, ls=style, label=label)

axs[0].axvspan(f_low, f_lf, color="grey", alpha=0.15)
axs[0].set(
    xlabel="Frequency [Hz]",
    ylabel=r"$S_{\Delta\phi}$ [rad$^2$/Hz]",
    title="Measured phase noise and hypothetical LF suppression",
    xlim=(f_low, f_high),
)
axs[0].legend(fontsize=10)

axs[1].axvspan(1.0 / f_lf, 1.0 / f_low,
               color="grey", alpha=0.15)
axs[1].axhline(1.0, color="black", ls="--", lw=0.9,
               label="Illustrative 1% phase-error budget")
axs[1].set(
    xlabel="Time between phase-reference updates, T [s]",
    ylabel="Estimated phase-induced error [%]",
    title="Gaussian phase model; frequency-cutoff approximation",
    xlim=(T.min(), T.max()),
)
axs[1].legend(fontsize=9)

for ax in axs:
    ax.grid(True, which="both", alpha=0.2)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import cumulative_trapezoid

# Run after your existing np.load lines.

def prepare_psd(f, S):
    f, S = np.asarray(f), np.asarray(S)

    valid = (
        np.isfinite(f) & np.isfinite(S)
        & (f > 0) & (S > 0)
    )
    f, S = f[valid], S[valid]

    order = np.argsort(f)
    f, S = f[order], S[order]

    if len(f) < 2 or np.any(np.diff(f) <= 0):
        raise ValueError(
            "Each PSD must contain distinct, increasing frequencies."
        )
    return f, S


datasets = [
    (OFF_OFF_x, OFF_OFF_y, "FSM OFF, PS OFF", "tab:blue", "-"),
    (ON_OFF_x, ON_OFF_y, "FSM ON, PS OFF", "tab:orange", "-"),
    (ON_ON_x, ON_ON_y, "FSM ON, PS ON", "tab:green", "-"),
    (OFF_ON_x, OFF_ON_y, "FSM OFF, PS ON", "tab:red", "-"),
    (D1_x, D1_y, "Detector-noise equivalent", "grey", ":"),
]

prepared = []
for f, S, label, colour, style in datasets:
    f, S = prepare_psd(f, S)
    prepared.append((f, S, label, colour, style))

# Common measured range, restricted to your original plotting range.
f_low = max(1.0, max(item[0].min() for item in prepared))
f_high = min(1e6, min(item[0].max() for item in prepared))

# IMPORTANT: lower f_high if the calibrated measurement bandwidth
# is smaller. A plotting limit alone does not establish validity.
if f_high <= 2 * f_low:
    raise ValueError("Insufficient common frequency range.")

# Union of measured frequency bins preserves resolved spectral peaks.
f_common = np.unique(np.concatenate(
    [item[0] for item in prepared] + [np.array([f_low, f_high])]
))
f_common = f_common[
    (f_common >= f_low) & (f_common <= f_high)
]

# Restrict T to the measured spectral range.
T = np.geomspace(2.0 / f_high, 1.0 / f_low, 600)


def interpolate_psd(f, S):
    return np.exp(np.interp(
        np.log(f_common), np.log(f), np.log(S)
    ))


def phase_variance(S):
    # Integral from each frequency to f_high.
    # Reverse integration avoids cancellation from subtracting totals.
    tail_integral = -cumulative_trapezoid(
        S[::-1], f_common[::-1], initial=0
    )[::-1]

    return np.interp(1.0 / T, f_common, tail_integral)


def phase_error_percent(variance):
    # 100 * (1 - exp(-variance/2)) / 2
    return -50.0 * np.expm1(-variance / 2.0)


plt.rcParams.update({
    "font.size": 11,
    "axes.titlesize": 14,
    "axes.labelsize": 13,
})

fig, axs = plt.subplots(
    2, 1,
    figsize=(8, 8),
    gridspec_kw={"height_ratios": [1.3, 1]},
    constrained_layout=True,
)

results = {}

for f, S, label, colour, style in prepared:
    S_common = interpolate_psd(f, S)
    variance = phase_variance(S_common)
    error = phase_error_percent(variance)

    results[label] = {
        "variance_rad2": variance,
        "phase_error_percent": error,
    }

    axs[0].loglog(
        f_common, S_common,
        color=colour, linestyle=style,
        linewidth=1.5, label=label,
    )
    axs[1].semilogx(
        T, error,
        color=colour, linestyle=style,
        linewidth=1.5, label=label,
    )

# Retain your selected frequency markers in the PSD panel.
for frequency in [16, 190, 3700]:
    if f_low < frequency < f_high:
        axs[0].axvline(
            frequency, color="black", linestyle="--",
            alpha=0.35, linewidth=0.9,
        )

axs[0].set(
    xlim=(f_low, f_high),
    xlabel="Frequency [Hz]",
    ylabel=r"$S_{\Delta\phi}$ [rad$^2$/Hz]",
    title="(a) Phase noise with FSM and phase stabilisation",
)
axs[0].legend(loc="best", fontsize=10)

# An illustrative engineering budget, not a security threshold.
axs[1].axhline(
    1.0, color="black", linestyle="--", linewidth=0.9,
    label="Illustrative 1% phase-error budget",
)

axs[1].set(
    xlim=(T.min(), T.max()),
    xlabel="Time between phase-reference updates, T [s]",
    ylabel="Estimated phase-induced error [%]",
    title="(b) Gaussian phase model; frequency-cutoff approximation",
)
axs[1].set_ylim(bottom=0)
axs[1].legend(loc="best", fontsize=9)

for ax in axs:
    ax.grid(True, which="both", alpha=0.2)

fig.savefig("PSCE_PSD_Phase_Error.pdf", bbox_inches="tight")
plt.show()
```

```python
import numpy as np
from scipy.integrate import trapezoid

def phase_error_from_psd(f, Pxx, T, f_high=None):
    """
    f: one-sided frequency array [Hz]
    Pxx: differential phase PSD [rad^2/Hz]
    T: interval between phase-reference updates [s]
    f_high: upper valid measurement frequency [Hz]

    Assumes zero-mean Gaussian residual phase.
    Uses the approximate lower cutoff f_low = 1/T.
    """
    f = np.asarray(f, dtype=float)
    Pxx = np.asarray(Pxx, dtype=float)

    if f.ndim != 1 or Pxx.shape != f.shape:
        raise ValueError("f and Pxx must be matching 1D arrays.")
    if not np.isfinite(T) or T <= 0:
        raise ValueError("T must be positive.")

    valid = (
        np.isfinite(f) & np.isfinite(Pxx)
        & (f > 0) & (Pxx >= 0)
    )
    f, Pxx = f[valid], Pxx[valid]
    order = np.argsort(f)
    f, Pxx = f[order], Pxx[order]

    if len(f) < 2 or np.any(np.diff(f) <= 0):
        raise ValueError("Need distinct frequency bins.")

    f_low = 1.0 / T
    f_high = f[-1] if f_high is None else float(f_high)

    if not f[0] <= f_low < f_high <= f[-1]:
        raise ValueError(
            "Integration limits must lie within the measured PSD."
        )

    # Include interpolated values at the exact integration limits.
    inside = (f > f_low) & (f < f_high)
    f_int = np.concatenate(([f_low], f[inside], [f_high]))
    S_int = np.interp(f_int, f, Pxx)

    # Integrate against LINEAR frequency, even for a log-log plot.
    variance = trapezoid(S_int, x=f_int)
    rms_phase = np.sqrt(variance)
    error = -0.5 * np.expm1(-variance / 2.0)

    return variance, rms_phase, error


# Example: 100 ms between phase-reference updates.
# Set f_high to your validated measurement bandwidth if necessary.
variance, rms, error = phase_error_from_psd(
    ON_ON_x, ON_ON_y, T=0.1
)

print(f"Phase variance: {variance:.6g} rad²")
print(f"RMS phase:      {rms:.6g} rad")
print(f"Phase error:    {100 * error:.4f}%")
```

```python
from pathlib import Path
import datetime as dt
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.ticker import FuncFormatter, ScalarFormatter


# ============================================================
# 1. Files and output
# ============================================================

COUNTS_FILE = Path("receiver_bg_counts.npy")
SECONDS_FILE = Path("receiver_bg_seconds.npy")
TIMESTAMPS_FILE = Path(
    r"D:\PhD Thesis Writing\Data\2025-09-24"
    r"\20Mhz_Background_24hour_24_09_2025ts.npy"
)

OUTPUT_DIR = Path("background_analysis_output")
SAVE_OUTPUTS = True

# Your thesis describes the stored values as counts per second.
INPUT_UNITS = "cps"       # "cps" or "counts"
BIN_DURATION_S = None    # Required only for "counts"


# ============================================================
# 2. Measurement configuration
# ============================================================

GATE_FREQUENCY_HZ = 20e6
GATE_WIDTH_S = 2.2e-9
MEASUREMENT_DUTY = GATE_FREQUENCY_HZ * GATE_WIDTH_S

DEAD_TIME_S = 15e-6

# Set False if acquisition software already corrected dead time.
APPLY_DEAD_TIME_CORRECTION = True

# Measured dark rate under the same acquisition settings.
DARK_MEASURED_CPS = 7.5
DARK_RANGE_CPS = (5.0, 10.0)

VOA_DB = 10.81
T_VOA = 10**(-VOA_DB / 10)

# Predict operation without the measurement VOA.
REMOVE_VOA_FOR_QKD = True


# ============================================================
# 3. QKD configuration
# ============================================================

# Chapter 2 assumptions.
QKD_REPETITION_HZ = 2e6
QKD_WINDOW_S = 1e-9
QKD_DUTY = QKD_REPETITION_HZ * QKD_WINDOW_S

SIGNAL_DV_CPS = 20.0
SIGNAL_TF_TOTAL_CPS = 20.0   # Total signal across both TF outputs.
SIGNAL_ERROR = 0.01

# TF-like model:
# Two identical receiving arms, balanced beamsplitter,
# two detectors, random background errors.
#
# This is an optical error-rate model, not a TF security analysis.

SHOW_DARK_RANGE = True


# ============================================================
# 4. Assumed radiance response
# ============================================================

# These are model assumptions, NOT a calibrated filter response.
WAVELENGTH_M = 1550e-9

BAND_MIN_NM = 1260.0
BAND_MAX_NM = 1625.0
BANDWIDTH_NM = BAND_MAX_NM - BAND_MIN_NM

DETECTOR_EFFICIENCY = 0.20
OPTICAL_TRANSMISSION = 0.80   # Excludes VOA and detector efficiency.

# Ideal single-spatial-mode collection, both polarizations accepted.
# Evaluate at the nominal wavelength throughout the assumed band.
EFFECTIVE_ETENDUE = WAVELENGTH_M**2

H_PLANCK = 6.62607015e-34
C_LIGHT = 299792458.0
PHOTON_ENERGY = H_PLANCK * C_LIGHT / WAVELENGTH_M


# ============================================================
# 5. Load and validate
# ============================================================

xs = np.asarray(
    np.load(COUNTS_FILE, allow_pickle=False), dtype=float
).squeeze()

ts = np.asarray(
    np.load(SECONDS_FILE, allow_pickle=False), dtype=float
).squeeze()

timestamps = np.asarray(
    np.load(TIMESTAMPS_FILE, allow_pickle=False)
).reshape(-1)

if xs.ndim != 1 or ts.ndim != 1:
    raise ValueError("Counts and elapsed times must be one-dimensional.")

if len(xs) != len(ts) or len(xs) < 2:
    raise ValueError("Counts and times must have matching lengths >= 2.")

if not np.all(np.isfinite(xs)) or not np.all(np.isfinite(ts)):
    raise ValueError("Counts or elapsed times contain non-finite values.")

if np.any(xs < 0) or np.any(np.diff(ts) <= 0):
    raise ValueError("Counts must be non-negative and times increasing.")

if timestamps.size == 0:
    raise ValueError("Timestamp file is empty.")

if INPUT_UNITS == "cps":
    measured_rate = xs.copy()
elif INPUT_UNITS == "counts":
    if BIN_DURATION_S is None or BIN_DURATION_S <= 0:
        raise ValueError("Supply a positive BIN_DURATION_S.")
    measured_rate = xs / BIN_DURATION_S
else:
    raise ValueError("INPUT_UNITS must be 'cps' or 'counts'.")

if not 0 < MEASUREMENT_DUTY <= 1:
    raise ValueError("Invalid measurement gate duty cycle.")

if not 0 < QKD_DUTY <= 1:
    raise ValueError("Invalid QKD acceptance duty cycle.")

if min(SIGNAL_DV_CPS, SIGNAL_TF_TOTAL_CPS) <= 0:
    raise ValueError("Signal count rates must be positive.")

if not 0 <= SIGNAL_ERROR <= 0.5:
    raise ValueError("Set SIGNAL_ERROR between 0 and 0.5.")

if min(DARK_RANGE_CPS) < 0 or DARK_MEASURED_CPS < 0:
    raise ValueError("Dark count rates must be non-negative.")

if BANDWIDTH_NM <= 0:
    raise ValueError("Assumed bandwidth must be positive.")

if not (
    0 < DETECTOR_EFFICIENCY <= 1
    and 0 < OPTICAL_TRANSMISSION <= 1
):
    raise ValueError("Efficiency and transmission must be in (0, 1].")

# Align the first elapsed-time sample with timestamps[0].
ts = ts - ts[0]

start_ts = timestamps[0]
if isinstance(start_ts, (bytes, np.bytes_)):
    start_ts = start_ts.decode()
else:
    start_ts = str(start_ts)

start_dt = dt.datetime.strptime(start_ts, "%H:%M:%S:%f")


# ============================================================
# 6. Dead-time correction and background conversion
# ============================================================

def linearise_rate(rate):
    """Approximate non-paralysable dead-time correction."""
    rate = np.asarray(rate, dtype=float)

    if np.any(~np.isfinite(rate)) or np.any(rate < 0):
        raise ValueError("Rates must be finite and non-negative.")

    if not APPLY_DEAD_TIME_CORRECTION:
        return rate

    occupation = rate * DEAD_TIME_S
    if np.any(occupation >= 1):
        raise ValueError("Count rate exceeds the dead-time model limit.")

    return rate / (1 - occupation)


if (
    APPLY_DEAD_TIME_CORRECTION
    and np.max(measured_rate) * DEAD_TIME_S > 0.1
):
    warnings.warn(
        "Dead-time correction exceeds 11% for some samples. "
        "Results depend on the assumed detector dead-time model."
    )

measured_linear = linearise_rate(measured_rate)


def calculate_background(dark_cps):
    dark_linear = float(linearise_rate(dark_cps))

    # Signed estimate retained for radiance.
    # Do not multiply detector dark counts by an optical-loss correction.
    sky_gated = measured_linear - dark_linear

    # For QBER, use a non-negative estimated optical background.
    sky_qkd = (
        np.maximum(sky_gated, 0)
        * QKD_DUTY / MEASUREMENT_DUTY
    )

    if REMOVE_VOA_FOR_QKD:
        sky_qkd = sky_qkd / T_VOA

    # Approximate dark scaling with active gate duration.
    # Prefer a measurement at the intended QKD settings when available.
    dark_qkd = dark_linear * QKD_DUTY / MEASUREMENT_DUTY

    # Chapter 2 model.
    noise_dv = sky_qkd + dark_qkd
    q_dv = (
        SIGNAL_ERROR * SIGNAL_DV_CPS + 0.5 * noise_dv
    ) / (SIGNAL_DV_CPS + noise_dv)

    # TF-like model:
    # Sum background from two arms and dark noise from two detectors.
    # A beamsplitter redistributes photons; it does not double them again.
    noise_tf = 2 * sky_qkd + 2 * dark_qkd
    q_tf = (
        SIGNAL_ERROR * SIGNAL_TF_TOTAL_CPS + 0.5 * noise_tf
    ) / (SIGNAL_TF_TOTAL_CPS + noise_tf)

    photon_radiance = sky_gated / (
        MEASUREMENT_DUTY
        * T_VOA
        * DETECTOR_EFFICIENCY
        * OPTICAL_TRANSMISSION
        * EFFECTIVE_ETENDUE
        * BANDWIDTH_NM
    )

    energy_radiance = photon_radiance * PHOTON_ENERGY

    return {
        "sky_qkd_cps": sky_qkd,
        "dark_qkd_cps": dark_qkd,
        "q_dv": q_dv,
        "q_tf": q_tf,
        "photon_radiance": photon_radiance,
        "energy_radiance": energy_radiance,
    }


result = calculate_background(DARK_MEASURED_CPS)

dark_results = [
    calculate_background(dark)
    for dark in DARK_RANGE_CPS
]

q_dv_low = np.minimum(
    dark_results[0]["q_dv"], dark_results[1]["q_dv"]
)
q_dv_high = np.maximum(
    dark_results[0]["q_dv"], dark_results[1]["q_dv"]
)

q_tf_low = np.minimum(
    dark_results[0]["q_tf"], dark_results[1]["q_tf"]
)
q_tf_high = np.maximum(
    dark_results[0]["q_tf"], dark_results[1]["q_tf"]
)


# ============================================================
# 7. Clock ticks and sky-state shading
# ============================================================

def clock_formatter(x, pos):
    real_time = start_dt + dt.timedelta(seconds=float(x))
    return real_time.strftime("%I %p").lstrip("0")


start_clock_s = (
    start_dt.hour * 3600
    + start_dt.minute * 60
    + start_dt.second
    + start_dt.microsecond / 1e6
)

tick_spacing = 3 * 3600
first_tick = (
    np.ceil(start_clock_s / tick_spacing) * tick_spacing
    - start_clock_s
)
ticks = np.arange(first_tick, ts[-1] + 1, tick_spacing)

SKY_COLOURS = {
    "Night": "#1f2d3a",
    "Astronomical Twilight": "#405463",
    "Nautical Twilight": "#6b8190",
    "Civil Twilight": "#a8c1cf",
    "Daylight": "#cfe7f2",
}

# Your supplied sky-state times, repeated over the plotting interval.
# These are not recalculated astronomical times.
SKY_INTERVALS = [
    ("Night", 0, 4 * 3600 + 53 * 60),
    ("Astronomical Twilight", 4 * 3600 + 53 * 60,
     5 * 3600 + 36 * 60),
    ("Nautical Twilight", 5 * 3600 + 36 * 60,
     6 * 3600 + 18 * 60),
    ("Civil Twilight", 6 * 3600 + 18 * 60,
     6 * 3600 + 53 * 60),
    ("Daylight", 6 * 3600 + 53 * 60,
     18 * 3600 + 57 * 60),
    ("Civil Twilight", 18 * 3600 + 57 * 60,
     19 * 3600 + 32 * 60),
    ("Nautical Twilight", 19 * 3600 + 32 * 60,
     20 * 3600 + 14 * 60),
    ("Astronomical Twilight", 20 * 3600 + 14 * 60,
     20 * 3600 + 57 * 60),
    ("Night", 20 * 3600 + 57 * 60, 24 * 3600),
]


def shade_sky_states(axis, xmax, legend=False, alpha=0.35):
    days = int(np.ceil((start_clock_s + xmax) / 86400)) + 1

    for day in range(-1, days):
        for label, begin, end in SKY_INTERVALS:
            x0 = day * 86400 + begin - start_clock_s
            x1 = day * 86400 + end - start_clock_s

            left = max(0, x0)
            right = min(xmax, x1)

            if right > left:
                axis.axvspan(
                    left, right,
                    facecolor=SKY_COLOURS[label],
                    edgecolor="none",
                    alpha=alpha,
                    zorder=0
                )

    if legend:
        patches = [
            mpatches.Patch(
                facecolor=colour, alpha=alpha, label=label
            )
            for label, colour in SKY_COLOURS.items()
        ]
        axis.legend(
            handles=patches,
            loc="upper right",
            fontsize=8,
            frameon=True,
            facecolor="white",
            framealpha=1
        )


# ============================================================
# 8. Plot
# ============================================================

plt.rcParams.update({'font.size': 10})
plt.rc('axes', titlesize=16)
plt.rc('axes', labelsize=14)

fig, ax = plt.subplots(
    2, 1,
    sharex=True,
    gridspec_kw={'height_ratios': [2, 1]}
)
fig.set_size_inches(8, 6)

ax[0].plot(
    ts,
    result["photon_radiance"],
    color="tab:blue",
    linewidth=1
)

ax[1].plot(
    ts, 100 * result["q_dv"],
    color="tab:blue",
    linewidth=1,
    label="Chapter 2 model"
)

if SHOW_DARK_RANGE:
    # Sensitivity to the assumed dark count rate, not confidence intervals.
    ax[1].fill_between(
        ts, 100 * q_dv_low, 100 * q_dv_high,
        color="tab:blue", alpha=0.18, linewidth=0
    )
    ax[1].fill_between(
        ts, 100 * q_tf_low, 100 * q_tf_high,
        color="tab:orange", alpha=0.18, linewidth=0
    )

ax[1].axhline(
    100 * SIGNAL_ERROR,
    linestyle="--",
    color="k",
    alpha=0.45,
    linewidth=0.95
)

ax[0].set_ylabel(
    r'[cts/(s m$^{2}$ sr nm)]' '\n'
    r'$L_\lambda$',
    fontsize=12
)
ax[1].set_ylabel("Est. Q [%]")
ax[1].set_xlabel("Time of day")

ax[0].set_title("Estimated Background Sky Radiance & QBER")
ax[1].set_ylim(0, 50)

formatter = ScalarFormatter(useMathText=True)
formatter.set_powerlimits((0, 0))
formatter.set_useOffset(False)
ax[0].yaxis.set_major_formatter(formatter)

for axis in ax:
    axis.set_xticks(ticks)
    axis.xaxis.set_major_formatter(FuncFormatter(clock_formatter))
    axis.grid(
        linestyle="--",
        color="black",
        alpha=0.25,
        axis="x"
    )

shade_sky_states(ax[0], xmax=ts[-1], legend=True)
shade_sky_states(ax[1], xmax=ts[-1], legend=False)

ax[1].legend(
    loc="best",
    fontsize=10,
    frameon=True,
    facecolor="white",
    framealpha=1
)

ax[0].tick_params(axis="x", labelbottom=False)
ax[1].set_xlim(0, ts[-1])

fig.subplots_adjust(
    left=0.16,
    bottom=0.12,
    hspace=0.045
)


# ============================================================
# 9. Export
# ============================================================

table = pd.DataFrame({
    "elapsed_s": ts,
    "measured_cps": measured_rate,
    "dead_time_corrected_cps": measured_linear,
    "equivalent_photon_radiance": result["photon_radiance"],
    "equivalent_energy_radiance_W_m2_sr_nm": result["energy_radiance"],
    "sky_qkd_cps_per_arm": result["sky_qkd_cps"],
    "dark_qkd_cps_per_detector": result["dark_qkd_cps"],
    "qber_chapter2_pct": 100 * result["q_dv"],
    "qber_tf_like_pct": 100 * result["q_tf"],
    "qber_chapter2_dark_range_low_pct": 100 * q_dv_low,
    "qber_chapter2_dark_range_high_pct": 100 * q_dv_high,
    "qber_tf_dark_range_low_pct": 100 * q_tf_low,
    "qber_tf_dark_range_high_pct": 100 * q_tf_high,
})

if SAVE_OUTPUTS:
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

    # No bbox_inches="tight": preserve the 8 × 6 inch canvas.
    fig.savefig(
        OUTPUT_DIR / "C5_Receiver_Radiance_QBER.pdf",
        dpi=300
    )
    fig.savefig(
        OUTPUT_DIR / "C5_Receiver_Radiance_QBER.png",
        dpi=300
    )
    table.to_csv(
        OUTPUT_DIR / "C5_Receiver_Radiance_QBER.csv",
        index=False
    )

    print(f"Saved outputs to: {OUTPUT_DIR.resolve()}")

print(f"Measurement duty cycle: {MEASUREMENT_DUTY:.3%}")
print(f"QKD acceptance duty cycle: {QKD_DUTY:.3%}")
print(f"Assumed étendue: {EFFECTIVE_ETENDUE:.4g} m² sr")
print(f"Assumed optical bandwidth: {BANDWIDTH_NM:g} nm")
print(
    "Radiance is response-model dependent. "
    "Dark-range shading is a sensitivity range, not a confidence interval."
)

plt.show()
```

```python
#C5: Background Count Spectrum 24H

lam_m = np.load('rec_bg_sum_binned_ls.npy')
counts = np.load('rec_bg_sum_binned_xs.npy')
sigma  = np.load('rec_bg_sum_binned_std.npy')

# Convert to nanometers for plotting
lam_nm = lam_m
bin_width = lam_m[1] - lam_m[0]  # ~1 nm

# --- Style settings ---
plt.rcParams.update({'font.size': 10})
plt.rc('axes', titlesize=18)
plt.rc('axes', labelsize=16)

fig, ax = plt.subplots()
fig.set_size_inches(8, 6)

# --- Light shaded bars (uniform shading) ---
ax.bar(lam_nm, counts, width=bin_width, align='center',
       color='tab:blue', alpha=0.25, edgecolor='none')
ax.step(lam_nm, counts, where='mid',
        color='tab:blue', lw=0.8, zorder=2)
# --- Error bars + small markers ---
ax.errorbar(lam_nm, counts, yerr=sigma,
            fmt='s', ms=3.5, color='tab:blue',
            ecolor='tab:blue', elinewidth=1.2, capsize=1.5,
            label='Counts $±1σ$')

# --- Axis limits: expand by half a bin width ---
xmin = lam_nm.min() - bin_width/2
xmax = lam_nm.max() + bin_width/2
ax.set_xlim(xmin, xmax)

# --- Major ticks every 10 nm, minor ticks every 1 nm ---
ax.xaxis.set_major_locator(MultipleLocator(10))
ax.xaxis.set_minor_locator(MultipleLocator(1))

def nm_formatter(x, pos):
    return f"{int(round(x))}"

ax.xaxis.set_major_formatter(FuncFormatter(nm_formatter))
ax.axvline(x=1.55e3, linestyle='--', color='k', alpha=0.55, linewidth = '1.2')
# --- Labels & formatting ---
ax.set_xlabel('Wavelength [nm]')
ax.set_ylabel('Counts [#]')
ax.set_title('Receiver Cumulative Background Spectrum, 24H')
ax.grid(True, which='major', alpha=0.35)
ax.legend(loc='best', fontsize=14)

plt.tight_layout()
plt.show()
plt.savefig('C5 Cumulative Background Spectrum.pdf', dpi=300)
```
