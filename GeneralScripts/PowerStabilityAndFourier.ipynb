```python
def generatenp(directory,files,taus):
    nps=[]
    for i in range(len(files)):
        print(files[i])
        nps.append((np.genfromtxt(r'{}{}'.format(directory,files[i]),skip_header=23,usecols=0 ,delimiter=',', names=True, case_sensitive=True),np.genfromtxt(r'{}{}'.format(directory,files[i]),skip_header=23,usecols=1 ,delimiter=',', names=True, case_sensitive=True),taus))
    return(nps)
```

```python
def fourierget(data,time):
    N=len(data)
    samplerate=N/time
    yfourier=fft(data)
    xfourier=fftfreq(N, 1/samplerate)
    return(xfourier,yfourier)
```

```python
fouriers=[]
for i in range(len(june28data)):
    fouriers.append(fourierget(june28data[i][1],499.99936))
    
```
